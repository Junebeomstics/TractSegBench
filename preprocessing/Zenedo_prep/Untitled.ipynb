```python
import nibabel as nib
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def load_nifti(file_path):
    """Load NIfTI file and return image data."""
    img = nib.load(file_path)
    return img.get_fdata()

def plot_slice(mask_data, slice_idx, axis=0):
    """Plot a single slice from the NIfTI mask data."""
    plt.figure(figsize=(6, 6))
    if axis == 0:
        plt.imshow(mask_data[slice_idx, :, :], cmap="gray", origin="lower")
    elif axis == 1:
        plt.imshow(mask_data[:, slice_idx, :], cmap="gray", origin="lower")
    else:
        plt.imshow(mask_data[:, :, slice_idx], cmap="gray", origin="lower")
    plt.axis("off")
    plt.title(f"Slice {slice_idx} (Axis {axis})")
    plt.show()

def interactive_viewer(file_path):
    """Create an interactive viewer for NIfTI mask slices."""
    mask_data = load_nifti(file_path)
    shape = mask_data.shape

    interact(plot_slice, 
             mask_data=widgets.fixed(mask_data),
             slice_idx=widgets.IntSlider(min=0, max=shape[2]-1, step=1, value=shape[2]//2),
             axis=widgets.Dropdown(options=[("Sagittal", 0), ("Coronal", 1), ("Axial", 2)], 
                                   value=2, description="Axis"))

# 사용 예시
nifti_mask_path = '/global/cfs/cdirs/m4673/junbeom/TractSegVis/TractSeg/preprocessing/Zenedo_prep/global_brain_mask.nii.gz'  # 실제 경로로 변경
interactive_viewer(nifti_mask_path)
```
