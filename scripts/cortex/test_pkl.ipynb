```python
import pickle
import numpy as np

# Load PKL file
def load_pkl_and_check_shape(file_path):
    try:
        # Load the PKL file
        with open(file_path, 'rb') as f:
            data = pickle.load(f)
        
        print(f"File loaded successfully: {file_path}")
        print(f"Data type: {type(data)}")
        
        # Check shape based on data type
        if hasattr(data, 'shape'):
            # numpy array, pandas DataFrame, etc.
            print(f"Shape: {data.shape}")
        elif isinstance(data, dict):
            print("Data is in dictionary form.")
            for key, value in data.items():
                if hasattr(value, 'shape'):
                    print(f"  {key}: {value.shape}")
                else:
                    print(f"  {key}: {type(value)} (length: {len(value) if hasattr(value, '__len__') else 'N/A'})")
        elif isinstance(data, list):
            print(f"List form - length: {len(data)}")
            if len(data) > 0 and hasattr(data[0], 'shape'):
                print(f"Shape of the first element: {data[0].shape}")
        elif isinstance(data, tuple):
            print(f"Tuple form - length: {len(data)}")
            for i, item in enumerate(data):
                if hasattr(item, 'shape'):
                    print(f"  element {i}: {item.shape}")
        else:
            print(f"Other type - length: {len(data) if hasattr(data, '__len__') else 'N/A'}")
            
        return data
        
    except Exception as e:
        print(f"Error occurred while loading file: {e}")
        return None

# Usage example
file_path = "/mnt/scratch/junb/TractSegVis/TractSeg/icosphere_2.pkl"  # Enter the file path here
data = load_pkl_and_check_shape(file_path)
```
