```python
import pickle
import numpy as np

# Load PKL file
def load_pkl_and_check_shape(file_path):
    try:
        # Load the PKL file
        with open(file_path, 'rb') as f:
            data = pickle.load(f)
        
        print(f"파일 로드 성공: {file_path}")
        print(f"데이터 타입: {type(data)}")
        
        # Check shape based on data type
        if hasattr(data, 'shape'):
            # numpy array, pandas DataFrame, etc.
            print(f"Shape: {data.shape}")
        elif isinstance(data, dict):
            print("딕셔너리 형태의 데이터입니다.")
            for key, value in data.items():
                if hasattr(value, 'shape'):
                    print(f"  {key}: {value.shape}")
                else:
                    print(f"  {key}: {type(value)} (길이: {len(value) if hasattr(value, '__len__') else 'N/A'})")
        elif isinstance(data, list):
            print(f"리스트 형태 - 길이: {len(data)}")
            if len(data) > 0 and hasattr(data[0], 'shape'):
                print(f"첫 번째 요소의 shape: {data[0].shape}")
        elif isinstance(data, tuple):
            print(f"튜플 형태 - 길이: {len(data)}")
            for i, item in enumerate(data):
                if hasattr(item, 'shape'):
                    print(f"  요소 {i}: {item.shape}")
        else:
            print(f"기타 타입 - 길이: {len(data) if hasattr(data, '__len__') else 'N/A'}")
            
        return data
        
    except Exception as e:
        print(f"파일 로드 중 오류 발생: {e}")
        return None

# Usage example
file_path = "/mnt/scratch/junb/TractSegVis/TractSeg/icosphere_2.pkl"  # Enter the file path here
data = load_pkl_and_check_shape(file_path)
```
